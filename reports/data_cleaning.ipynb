```python
import pandas as pd
import ast
```

```python
games = pd.read_csv('../data/processed/games_with_ratings.csv')
```

```python
list_columns = ['categories', 'genres', 'supported_languages', 'tags']

games['supported_languages'] = games['supported_languages'].apply(
    lambda x: x.replace("K'iche'", '"K\'iche"') if isinstance(x, str) else x
)

for col in list_columns:
    games[col] = games[col].apply(lambda x: ast.literal_eval(x) if isinstance(x, str) else x)

games['categories'] = games['categories'].apply(lambda x: ['No categories'] if x == [''] else x)
games['tags'] = games['tags'].apply(lambda x: ['No tags'] if x == [''] else x)
games['genres'] = games['genres'].apply(lambda x: ['No genres'] if x == [''] else x)
```

```python
import re

def clean_language_string(language):
    language = re.sub(r'&[a-zA-Z]+;', '', language)
    language = re.sub(r'\r\n|\n|\r', '', language)
    language = re.sub(r'\s+', ' ', language)

    if 'Japanese' in language:
        return 'Japanese'
    elif 'Chinese' in language:
        if 'Simplified' in language:
            return 'Simplified Chinese'
        elif 'Traditional' in language:
            return 'Traditional Chinese'
    elif 'English' in language:
        return 'English'
    elif 'French' in language:
        return 'French'
    elif 'German' in language:
        return 'German'
    elif 'Spanish' in language:
        return 'Spanish'
    elif 'Italian' in language:
        return 'Italian'
    elif 'Korean' in language:
        return 'Korean'
    elif 'Portuguese' in language:
        return 'Portuguese'
    elif 'Russian' in language:
        return 'Russian'
    elif 'Arabic' in language:
        return 'Arabic'
    elif 'Hindi' in language:
        return 'Hindi'
    elif 'Bengali' in language or 'Bangla' in language:
        return 'Bengali'
    elif 'Punjabi' in language:
        return 'Punjabi'
    else:
        return language

games['supported_languages'] = games['supported_languages'].apply(
    lambda x: [clean_language_string(lang) for lang in x] if isinstance(x, list) and x else x
)
```

```python
import os
import numpy as np

games['first_screenshot'] = games['screenshots'].apply(
    lambda x: x.split(',')[0] if isinstance(x, str) and ',' in x else x
)
```
